```python
#2 items × ₹500 = ₹1000 ,  If subtotal_idr says ₹1000, correct If it says ₹900, there is a data problem
order_items["calculated_subtotal"] = (
    order_items["quantity"] * order_items["unit_price_idr"]
)

print(
    "Subtotal mismatches:",
    len(
        order_items[
            order_items["subtotal_idr"] !=
            order_items["calculated_subtotal"]
        ]
    )
)
```

```python
#used strip() to remove extra spaces and title() to standardize text into proper capitalization.
products["category"] = (
    products["category"]
    .str.strip()
    .str.title()
)
```

```python
products["category"].value_counts()
```

```python
# Clean customer city and province
customers["city"] = customers["city"].str.strip().str.title()
customers["province"] = customers["province"].str.strip().str.title()

# Clean shipping city and province
orders["shipping_city"] = orders["shipping_city"].str.strip().str.title()
orders["shipping_province"] = orders["shipping_province"].str.strip().str.title()

# Clean order status
orders["order_status"] = orders["order_status"].str.strip().str.lower()
```

```python
#“I created year, month, month number, and year-month features from the order date for time-based analysis.”
orders["year"] = orders["order_date"].dt.year
orders["month"] = orders["order_date"].dt.month_name()
orders["month_num"] = orders["order_date"].dt.month

#It creates a Year-Month column like 2024-01 from order_date so we can easily analyze monthly trends.
orders["year_month"] = (
    orders["order_date"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
'''If order is cancelled → is_cancelled = 1, otherwise 0
If order is returned → is_returned = 1, otherwise 0
If order is delivered → is_delivered = 1, otherwise 0
“I created status flag columns using 1 and 0 so I could easily calculate order-status counts and rates.”'''
orders["is_cancelled"] = (orders["order_status"] == "cancelled").astype(int)
orders["is_returned"] = (orders["order_status"] == "returned").astype(int)
orders["is_delivered"] = (orders["order_status"] == "delivered").astype(int)
```

```python
'''
1 = discount used
0 = no discount
'''
orders["used_discount"] = (
    orders["discount_amount_idr"] > 0
).astype(int)
```
